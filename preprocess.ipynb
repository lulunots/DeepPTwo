```python
% %bash
set -ex

ome_folder="/working/lab_quann/louiseN/BC_data/Xenium/40xHE" 
vsi_folder="/working/lab_quann/louiseN/BC_data/Xenium/he"
data_folder="/working/lab_quann/louiseN/benchmark/deeppt/Xenium_slides_data" # compatible with deeppt pipeline naming conventions

for file in "$ome_folder"/*; do
    base=$(basename "$file" .ome.tif)
    vips tiffsave "$file" "${data_folder}/${base}.tif" --tile --pyramid --bigtiff --compression jpeg --Q 90 --tile-width 256 --tile-height 256 --strip
done

for entry in "$vsi_folder"/*.vsi; do
    base=$(basename "$entry" .vsi)
    if [ -d "$entry" ]; then
        file="$entry/$(basename "$entry")"
    else
        file="$entry"
    fi    
    bfconvert -tilex 256 -tiley 256 -pyramid-resolutions 6 -pyramid-scale 2 -compression JPEG-2000 "$file" "${data_folder}/${base}.ome.tif"
    vips tiffsave "${data_folder}/${base}.ome.tif" "${data_folder}/${base}.tif" --tile --pyramid --bigtiff --compression jpeg --Q 90 --tile-width 256 --tile-height 256 --strip
    rm "${data_folder}/${base}.ome.tif"
done
```
